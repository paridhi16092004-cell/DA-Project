```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

x = new['name']
y = new['marks']
z = new['subject ']

c = [
    "red", "blue", "green", "yellow", "orange",
    "purple", "pink", "cyan", "magenta", "lime",
    "teal", "gold", "black", "white"
]

# CORRECTED: Use keyword arguments for the star marker and edge color
plt.plot(x, y,
         markerfacecolor = "red",
         color='black',             # 'color' becomes color
         marker='o',               # 'style star' becomes marker='*'
         markeredgecolor='black',  # 'edecolor' becomes markeredgecolor
         linestyle='-',             # Leave empty if you only want the stars, or use '-' for a line
         markersize=10)            # Optional: makes the stars bigger
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# CORRECTED: plt.margins() is a separate function call, not an argument for plot()
plt.margins(0.05)

plt.bar(x, y,
        label=z,
        color=c,
        edgecolor='black',
        linewidth=1.2,
        width=0.8)

plt.xlabel('name')
plt.ylabel('marks')
plt.title('bar graph')
plt.legend()
plt.show()
```
